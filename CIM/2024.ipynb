## Question 1


```r
# (a) If there is a 0.30 probability that a child exposed to a certain contagious disease will catch the
# disease , what is the probability that,
# (i) out of 12 randomly selected children who were exposed to the disease, exactly 4 will catch
# the disease.
# [10 marks]
dbinom(4, size = 12, prob = 0.30)

# (ii) Out of 12 randomly selected children who were exposed to the disease, at least 3 will
# catch the disease?
# [10 marks]
dbinom(0, size = 12, prob = 0.30) + dbinom(1, size = 12, prob = 0.30) + dbinom(2, size = 12, prob = 0.30) + dbinom(3, size = 12, prob = 0.30)
pbinom(3, size = 12, prob = 0.30)

# (b) A trainee sharpshooter misses a target 60% of the time. Define the random variable X as
# the number of targets he misses before the first success.
# Find the probability that his first success is at the fourth try.
# [20 marks]
dgeom(3, prob = 0.4)

# (c) A machine is set to produce iron nails of length 2cm. A sample of 10 nails was taken and
# their lengths were measured in cm. The following results were obtained.
# 1.98, 1.99, 1.99, 2.00, 2.01, 2.02, 1.97, 1.96, 1.98, 1.99
# (i) Enter data into R and find the sample mean length of a nail and the sample variance.
# [10 marks]
nail_lengths <- c(1.98, 1.99, 1.99, 2.00, 2.01, 2.02, 1.97, 1.96, 1.98, 1.99)
mu <- mean(nail_lengths)
v <- var(nail_lengths)

# (ii) Construct a symmetrical 95% confidence interval for the mean length of a nail.
# [20 marks]
# ME = qt(0.975, df = length(nail_lengths) - 1) * sqrt(v / length(nail_lengths))
# c(mu-ME, mu+ME)
t.test(nail_lengths, conf.level = 0.95)

# (iii) Using confidence interval constructed in part (ii) or otherwise test at 5% significance
# level, whether the machine is in good working order.
# [30 marks]
t.test(nail_lengths, mu=2, conf.level = 0.95)
```

```r
# (a) The following table shows the assessed values and the selling prices of eight houses,
# constituting a random sample of all the houses sold recently in a metropolitan area.
# | Assessed Value (in ten thousand LKR) | Selling Price (in ten thousand LKR) |
# |--------------------------------------|--------------------------------------|
# | 70.3                                 | 114.4                                |
# | 102.0                                | 169.3                                |
# | 62.5                                 | 146.2                                |
# | 74.8                                 | 125.0                                |
# | 57.9                                 | 99.8                                 |
# | 81.6                                 | 132.1                                |
# | 110.4                                | 174.2                                |
# | 88.0                                 | 143.5                                |
# (i) Comment on the linearity of the assessed value of a house and the selling price using a suitable plot. [10 marks]
par(bg='white')
assessed_value <- c(70.3, 102.0, 62.5, 74.8, 57.9, 81.6, 110.4, 88.0)
selling_price <- c(114.4, 169.3, 146.2, 125.0, 99.8, 132.1, 174.2, 143.5)
plot(assessed_value, selling_price, 
        xlab = "Assessed Value (in ten thousand LKR)", 
        ylab = "Selling Price (in ten thousand LKR)", 
        main = "Scatter Plot of Assessed Value vs Selling Price")

abline(lm(selling_price ~ assessed_value))
```

```r
assessed_value < -c(70.3, 102.0, 62.5, 74.8, 57.9, 81.6, 110.4, 88.0)
selling_price < -c(114.4, 169.3, 146.2, 125.0, 99.8, 132.1, 174.2, 143.5)
# (ii) Calculate the Pearson’s correlation coefficient and Spearman’s rank correlation
# coefficient between the two variables and comment on that.
# [10 marks]
c(
    cor(assessed_value, selling_price, method="pearson"),
    cor(assessed_value, selling_price, method="spearman"),
)

# (iii) Re-calculate the Pearson’s correlation coefficient by removing any outlier observations
# Compare the value obtained with the previous value and explain the results. [10 marks]
selling_price < -selling_price[-3]
assessed_value < -assessed_value[-3]
c(
    cor(assessed_value, selling_price, method="pearson"),
    cor(assessed_value, selling_price, method="spearman"),
)

# b) An insurance company introduced two new health insurance policies, HP1 and HP2.They
# conducted presentations on HP1 and HP2 at Company A and Company B, respectively. Out
# of 254 employees of company B who attended the presentation, 136 decided to go for
# HP1.The presentation, which was held at Company B, attended by 175 employees and 63 out
# of them decided to go for HP2. The company wants to determine which health policy is more
# popular among the employees.
# (i) Estimate the (difference between two proportions of employees who decided to buy two health policies. [10 marks]
hp1 < -136 / 254
hp2 < -63 / 175
hp_diff < -hp1 - hp2
hp_diff

# (ii) Construct a symmetrical 95% Confidence Interval for the difference between two proportions. [20 marks]
me < -qnorm(0.975) * sqrt((hp1 * (1 - hp1) / 254) + (hp2 * (1 - hp2) / 175))
c(hp_diff - me, hp_diff + me)

# (iii) Test the hypothesis that the proportion of employees who decided to go for HP1 is the same as that for HP2. [10 marks]
prop.test(x=c(136, 63), n=c(254, 175), correct=FALSE)

# (c) A large manufacturing company receives raw materials from three different vendors and the
# data pertaining to the shipments received by the company from three vendors are given in the
# table below.
# | Vendor   | Number Rejected | Number Acceptable | Number Perfect |
# |----------|------------------|--------------------|----------------|
# | Vendor A | 12               | 23                 | 87             |
# | Vendor B | 10               | 26                 | 15             |
# | Vendor C | 21               | 30                 | 120            |
# (i) Define the hypotheses correctly to test whether the three vendors’ products are of equal quality. [10 marks]
"H_0: The proportions of acceptable products from all three vendors are equal."
"H_1: At least one vendor has a different proportion of acceptable products."

# (ii) Enter data suitably into R and perform a chi-square test to test the hypothesis defined
# in part (i) and comment on the results. [20 marks]
vendor_data < -matrix(c(12, 23, 87, 10, 26, 15, 21, 30, 120), nrow=3, byrow=TRUE)
rownames(vendor_data) < -c("Vendor A", "Vendor B", "Vendor C")
colnames(vendor_data) < -c("Number Rejected", "Number Acceptable", "Number Perfect")
chisq.test(vendor_data)
```
