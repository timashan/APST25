```r
par(bg="white")
df <- data.frame(a=c(1, 2, 3), b=c(3,2,1))
rownames(df) <- c("A", "B", "C")
colnames(df) <- c("A", "B")
chisq.test(df)
cor(df$A, df$B, method="pearson")
plot(df)
abline(lm(df))
```
