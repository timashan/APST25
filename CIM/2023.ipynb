## Question 02


```r
# a) The shortest time recorded by an athlete in a 400m race is called their personal best (PB).
# The PBs of the athletes in a large athletics club are normally distributed with mean 49.2 seconds and standard deviation 2.8 seconds.

# (i) Find the probability that a randomly chosen athlete from this club has a PB greater than 46 seconds. [20 marks]
1 - pnorm(46, mean=49.2, sd=2.8)

# (ii) Three athletes from the club are chosen at random. Find the probability that exactly 2 have PBs of less than 46 seconds. [30 marks]
dbinom(2, 3, pnorm(46, mean=49.2, sd=2.8))
```

```r
# b) In a very small empirical study a sample from a random variable X is observed.
# (1.28, 0.54, 0.56, 1.36, 0.55, 0.37, 0.43, 0.46, 0.62, 0.54, 0.54, 0.51, 0.44, 0.68, 0.55, 0.30)

# (i) Estimate the expected value of X.
x = c(
    1.28,
    0.54,
    0.56,
    1.36,
    0.55,
    0.37,
    0.43,
    0.46,
    0.62,
    0.54,
    0.54,
    0.51,
    0.44,
    0.68,
    0.55,
    0.30,
)
mean_x = mean(x)
mean_x

# (ii) Calculate the sample standard deviation.
sd(x)

# (iii) Check the normality of the X using Shapiro Wilk Normality test and write p-value obtained.
shapiro.test(x)

# (iv) Calculate a 95% confidence interval for the expected value of X assuming that X is Normally distributed. [50 marks]
e = qt(0.975, df=length(x) - 1) * sd(x) / sqrt(length(x))
c(mean_x - e, mean_x + e)
```

```r
# (a) Below is a part of a data set named ProductData.xlsx. Import the data set into R- Studio.
# | ProductID | Product | Wattage | Life Hours | Brand   | Unit Cost (Rs) |
# |-----------|---------|---------|------------|---------|----------------|
# |     1     | Bulb    |   200   |    3000    | Horizon |      450       |
# |     2     | Neon    |   100   |    2000    | Horizon |      250       |
# |     3     | Spot    |    60   |    4000    | Horizon |      500       |
# |     4     | Neon    |    10   |    8000    | Sunbeam |       80       |
# |     5     | Bulb    |    80   |    1000    | Horizon |      260       |
# |     6     | Spot    |   100   |    1500    | Horizon |      175       |
# |     7     | Spot    |   200   |    3000    | Horizon |      250       |
# |     8     | Spot    |    25   |    2000    | Sunbeam |      500       |
# |     9     | Bulb    |   200   |    3000    | Sunbeam |      550       |
# |    10     | Neon    |   100   |    2000    | Sunbeam |      180       |
# (i) Write the R command that you can use to see the number of data records and
# variables in the data-frame. [10 marks]

ProductData <- data.frame(
  ProductID = 1:10,
  Product = c("Bulb", "Neon", "Spot", "Neon", "Bulb", "Spot", "Spot", "Spot", "Bulb", "Neon"),
  Wattage = c(200, 100, 60, 10, 80, 100, 200, 25, 200, 100),
  Life_Hours = c(3000, 2000, 4000, 8000, 1000, 1500, 3000, 2000, 3000, 2000),
  Brand = c("Horizon", "Horizon", "Horizon", "Sunbeam", "Horizon", "Horizon", "Horizon", "Sunbeam", "Sunbeam", "Sunbeam"),
  Unit_Cost_Rs = c(450, 250, 500, 80, 260, 175, 250, 500, 550, 180)
)

# (ii) Suppose the wattage value with the ProductID equals 4 is entered incorrectly and
# it should be corrected as 100, write a R command to make the correction. [05 marks]
ProductData[4, "Wattage"] <- 4

# (iii) Calculate the average life hours for the two brands Horizon and Sunbeam. [05 marks]
tapply(ProductData$Life_Hours, ProductData$Brand, mean)

# (iv) Test at 5% significance level whether the life hours are different for the brands
# Horizon and Sunbeam (Clearly define your Hypotheses and interpret the output).
t.test(Life_Hours ~ Brand, data = ProductData)

# (v) Test at 5% significance level whether the life hours are different for the different
# products (Clearly define your Hypotheses and interpret the output). [15 marks]
anova_result <- aov(Life_Hours ~ Product, data = ProductData)
summary(anova_result)
```
