```r
df <- data.frame(Obs= 1:14, 
    Minutes=c(23, 29, 49, 64, 74, 87, 96, 97, 109, 119, 149, 145, 154, 166), 
    Units=c(1, 2, 3, 4, 4, 5, 6, 6, 7, 8, 9, 9, 10, 10))
df
```

```r
cor(df$Minutes, df$Units)
cor(df$Minutes, df$Units, method = "spearman")
```

```r
# lm(response~explanatory)
model = lm(Minutes~Units, data = df)
model
summary(model)
```
