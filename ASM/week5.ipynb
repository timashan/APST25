```r
salary_data<-read.csv("salary-survey.csv")
# Convert Education and Management to factors
salary_data$Education <- factor(salary_data$E,  levels = c(3, 2, 1), labels = c("Advanced", "Bachelor", "HighSchool"))
salary_data$Management <- factor(salary_data$M,levels = c(0, 1), labels = c("NonManager", "Manager"))
# # Main effects model
model_main <- lm(S ~ X + Education + Management + Education * Management, data = salary_data)
summary(model_main)
```

```r
# Simulated data with replicates
x <- rep(1:4, each = 3)
y <- c(3, 3.1, 2.9, 4.5, 4.7, 4.6, 6.2, 6.1, 6.3, 8, 7.9, 8.1)
# Linear model (restricted model)
model_linear <- lm(y ~ x)
# Saturated model (separate mean per level of x)
model_full <- lm(y ~ as.factor(x))
# Extract RSS (Residual Sum of Squares)
rss_linear <- deviance(model_linear) # total residuals
rss_full <- deviance(model_full) # pure error only
# Degrees of freedom
df_linear <- df.residual(model_linear)
df_full <- df.residual(model_full)
# Calculate SSLOF and SSPE
SSLOF <- rss_linear - rss_full
SSPE <- rss_full
df_LOF <- df_linear - df_full
df_PE <- df_full
# Mean Squares
MS_LOF <- SSLOF / df_LOF
MS_PE <- SSPE / df_PE
# F-statistic
F_value <- MS_LOF / MS_PE
p_value <- pf(F_value, df_LOF, df_PE, lower.tail = FALSE)
# Print results
cat("Lack of Fit SS =", SSLOF, "\n")
cat("Pure Error SS =", SSPE, "\n")
cat("F =", F_value, "\n")
cat("p-value =", p_value, "\n")
anova(model_linear, model_full)
```
