```r
data<-read.csv("Performance_data.csv")
fm <- lm(Y ~ X1 + X2 + X3 + X4 + X5 + X6, data = data)
summary(fm)
```

```r
rm <- lm(Y ~ X1 + X3 + X6, data = data)
summary(rm)
```

```r
anova(rm, fm)
```

```r
library(MASS)
stepAIC(fm, direction = "both")
```

```r
salary_data<-read.csv("salary-survey.csv")
# Convert Education and Management to factors
salary_data$Education <- factor(salary_data$E,  levels = c(3, 2, 1), labels = c("Advanced", "Bachelor", "HighSchool"))
salary_data$Management <- factor(salary_data$M,levels = c(0, 1), labels = c("NonManager", "Manager"))
# # Main effects model
model_main <- lm(S ~ X + Education + Management, data = salary_data)
summary(model_main)
```
