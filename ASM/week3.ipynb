```r
df <- data.frame(obs= 1:14, 
    minutes=c(23, 29, 49, 64, 74, 87, 96, 97, 109, 119, 149, 145, 154, 166), 
    units=c(1, 2, 3, 4, 4, 5, 6, 6, 7, 8, 9, 9, 10, 10))
df
```

```r
par(bg="white")
model = lm(minutes~units, data = df)
plot(df$units, df$minutes, main="Service Call Data",xlab="Units", ylab="Minutes")
abline(model, col="red", lwd=2)
abline(h=mean(df$minutes), col="blue", lty=2, lwd=2)
```

```r
m<-anova(model)
m
summary(model)
```

```r
f_stat <- summary(model)$fstatistic[1]
p_value <- pf(f_stat, 1, length(df$minutes)-2, lower.tail = FALSE)
f_stat
p_value
```

```r
options(repr.plot.width=15, repr.plot.height=10)
par(bg="white")
# Quick diagnostic plots
par(mfrow=c(2,2)) # request all 4 plots by 2*2
plot(model)
# Manual residual analysis
residuals <- resid(model)
fitted_values <- fitted(model)
# Check for patterns
# plot(fitted_values, residuals, main="Residuals vs Fitted", xlab="Fitted Values", ylab="Residuals")
```
